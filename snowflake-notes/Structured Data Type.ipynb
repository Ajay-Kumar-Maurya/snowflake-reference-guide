Structured data type =>
Similar to semi-structured data type, there are three structured data types in snowflake.
1. Structured array, 2. structured object, 3. structured map
There is no variant data type exists in structured data type, because variant does not have any schema.

In documentation, it is given how to convert structured data type to semi-structured data type and vice versa.
