# Programmatic Access Token (PAT) in Snowflake

For Snowflake authentication, any user must have information about:
- a) Account Identifier (account name)
- b) Username
- c) Password or PAT (Programmatic Access Token)

---

## What is a Programmatic Access Token?

In Snowflake, **PAT** stands for **Programmatic Access Token**.

A PAT is an authentication token that allows applications, scripts, APIs, CI/CD pipelines, and other automated processes to connect to Snowflake without using a username and password directly.

---

## Why Use PAT?

Instead of using **Username + Password** for authentication, you can use **Username + Programmatic Access Token** for authentication.

### Important Points:
- PAT does not replace the username; it replaces the password credential.
- The PAT is still associated with a Snowflake user.
- Snowflake validates the token and identifies the user it belongs to.
- PATs can be expired, rotated, and revoked independently of the user's password.

### This Provides:
- Better security
- Easier credential management
- Reduced password exposure
- Support for automation and integrations
- Short-lived credentials that can be revoked

---

## Common Use Cases

### 1. Python Applications

```python
import snowflake.connector
conn = snowflake.connector.connect(
    account="myaccount",
    user="john",
    token="<PAT>",
    authenticator="programmatic_access_token"
)
```

### 2. CI/CD Pipelines

Tools like:
- GitHub Actions
- Azure DevOps
- Jenkins
- GitLab CI

can authenticate to Snowflake using PATs instead of storing user passwords.

### 3. SnowSQL

```bash
snowsql -a myaccount -u myuser --authenticator programmatic_access_token
```

### 4. REST APIs and Drivers

PATs can be used by:
- JDBC
- ODBC
- Python Connector
- Snowflake CLI
- Custom applications

## Benefits over Password Authentication

| Feature | Password | PAT |
|---------|----------|-----|
| Lifespan | Long-lived | Can be short-lived |
| Rotation | Hard to rotate | Easy to regenerate |
| Scope | Shared among applications | Create separate token per application |
| Security | Higher security risk | More secure |
| Auditing | Cannot easily track usage | Better auditing |

---

## PAT Lifecycle

1. User creates PAT.
2. Application stores PAT securely.
3. Application authenticates using PAT.
4. PAT expires or is revoked.
5. New PAT is generated when needed.

---

## Example Scenario

Suppose an Azure Data Factory pipeline loads data into Snowflake every night.

**Without PAT:**
```
ADF → Username + Password → Snowflake
```

**With PAT:**
```
ADF → Username + PAT → Snowflake
```

If the PAT is compromised, you can revoke just that token without changing the user's password or affecting other applications.

---

## Difference Between PAT and OAuth

| Feature | PAT | OAuth |
|---------|-----|-------|
| Setup | Simpler setup | More complex setup |
| Token source | User-created token | Authorization-server based |
| Best for | Scripts and automation | Enterprise SSO applications |
| Delegation | Limited delegation features | Rich delegation and consent model |

---

## When to Use PAT?
- ✅ Scripts and automation
- ✅ CI/CD pipelines
- ✅ Data engineering jobs
- ✅ Service integrations

## When to Use OAuth?
- ✅ Enterprise applications
- ✅ User sign-in via SSO (Azure AD/Entra ID, Okta, etc.)
- ✅ Fine-grained delegated access

---

## Summary

PAT (Programmatic Access Token) is a secure token-based authentication mechanism in Snowflake that enables applications and automated processes to access Snowflake without using passwords directly.
